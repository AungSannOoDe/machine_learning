```python
import pandas as pd
import numpy as np
from sklearn.metrics import log_loss,accuracy_score,confusion_matrix, f1_score,classification_report
from sklearn.preprocessing import StandardScaler,MinMaxScaler
from  sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from  sklearn.pipeline import  Pipeline
```

```python
k_df=pd.read_csv("../Datasets/Kyphosis.csv")
```

```python
print(k_df.info())
```

```python
X=k_df.drop(columns=["Kyphosis"])
y=k_df["Kyphosis"]
```

```python
print(y.value_counts())
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,y,test_size=0.3,random_state=25,stratify=y)
```

```python
# SVM ko thoung kernel ka linear
svm=SVC(kernel="linear",C=1)
svm.fit(X_train,Y_train)
y_pred=svm.predict(X_test)
print(f"F-1 score :{f1_score(Y_test,y_pred,pos_label="present")}")
print(f"Classification Report: \n {classification_report(Y_test,y_pred)}")
```

```python
c_s=np.linspace(0.001,5,20)
score=list()
for c in c_s:
    svm=SVC(C=c,kernel="linear")
    svm.fit(X_train,Y_train)
    y_pred=svm.predict(X_test)
    score.append([c,f1_score(Y_test,y_pred,pos_label="present")])
score_df=pd.DataFrame(score,columns=["C","f1_score"]).sort_values("f1_score",ascending=False)
print(score_df)
```

```python
scaler=StandardScaler()
x_train_scaler=scaler.fit_transform(X_train)
x_test_scaler=scaler.transform(X_test)
```

```python
c_s=np.linspace(0.001,5,20)
score=list()
for c in c_s:
    svm=SVC(C=c,kernel="linear")
    pipe=Pipeline([("SCL",scaler),("SVM",svm)])
    pipe.fit(x_train_scaler,Y_train)
    y_pred=pipe.predict(x_test_scaler)
    score.append([c,f1_score(Y_test,y_pred,pos_label="present")])
score_df=pd.DataFrame(score,columns=["C","f1_score"]).sort_values("f1_score",ascending=False)
print("Top 5 Result")
print(score_df.head(5))
```

```python
min_max_scaler=MinMaxScaler()
X_train_min=min_max_scaler.fit_transform(X_train)
X_test_min=min_max_scaler.transform(X_test)
```

```python
c_s=np.linspace(0.001,5,20)
score=list()
for c in c_s:
    svm=SVC(C=c,kernel="linear")
    pipe=Pipeline([("SCL",min_max_scaler),("SVM",svm)])
    pipe.fit(X_train_min,Y_train)
    y_pred=pipe.predict(X_test_min)
    score.append([c,f1_score(Y_test,y_pred,pos_label="absent")])
score_df=pd.DataFrame(score,columns=["C","f1_score"]).sort_values("f1_score",ascending=False)
print("Top 5 Result")
print(score_df)
```
