```python
import pandas as pd
import numpy as np
from sklearn.metrics import log_loss,accuracy_score,confusion_matrix, f1_score,classification_report
from sklearn.preprocessing import StandardScaler,MinMaxScaler
from  sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from  sklearn.pipeline import  Pipeline
```

```python
b_df=pd.read_csv("../Datasets/BreastCancer.csv")
```

```python
svm = SVC(kernel='rbf', gamma=0.5 , C=1)
```

```python
X=b_df.drop(columns=["Class"])
Y=b_df["Class"]
```

```python
from sklearn.preprocessing import LabelEncoder

# Initialize the encoder
le = LabelEncoder()

# Fit and transform your target column
Y = le.fit_transform(b_df["Class"])
```

```python
print(Y.unique())
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=25,stratify=Y)
```

```python
svm.fit(X_train,Y_train)
y_pred=svm.predict(X_test)
print(f1_score(Y_test,y_pred,pos_label="Malignant"))
print("Accuracy:",accuracy_score(Y_test,y_pred,))
print("Confusion Matrix:")
print(confusion_matrix(Y_test,y_pred))
print("Classification Report:")
print(classification_report(Y_test,y_pred))
```

```python
c_s=np.linspace(0.001,5,20)
g_s=np.linspace(0.001,5,20)
acc_score=[]
for xc in c_s:
    for gs in g_s:
        svm=SVC(kernel='rbf', gamma=xc, C=gs)
        svm.fit(X_train,Y_train)
        y_pred=svm.predict(X_test)
        acc_score.append([xc,gs,f1_score(Y_test,y_pred,pos_label="Malignant")])
score_df=pd.DataFrame(acc_score,columns=["C","Gemma","F1"]).sort_values(by="F1",ascending=False)
print(score_df)
```

```python
svm=SVC(kernel='poly',C=1,degree=4)
svm.fit(X_train,Y_train)
y_pred=svm.predict(X_test)
print(f1_score(Y_test,y_pred,pos_label="Malignant"))
print("Accuracy:",accuracy_score(Y_test,y_pred))
print("Confusion Matrix:")
print(confusion_matrix(Y_test,y_pred))
print("Classification Report:")
print(classification_report(Y_test,y_pred))
```
