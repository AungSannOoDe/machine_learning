```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
from sklearn.metrics import log_loss,accuracy_score,confusion_matrix, f1_score,classification_report
from sklearn.preprocessing import StandardScaler,MinMaxScaler
from  sklearn.model_selection import train_test_split
from tqdm import tqdm
from sklearn.svm import SVC
```

```python
df=pd.read_csv("../Datasets/Glass.csv")
```

```python
df.columns
```

```python
X=df.drop(columns=["Type"])
Y=df["Type"]
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=25,stratify=Y)
```

```python
svm=SVC(kernel="linear",C=1,decision_function_shape="ovr")
svm.fit(X_train,Y_train)
y_pred=svm.predict(X_test)
print(classification_report(Y_test,y_pred))
```

```python
svm=SVC(kernel="linear",C=1,decision_function_shape="ovo")
svm.fit(X_train,Y_train)
y_pred=svm.predict(X_test)
print(classification_report(Y_test,y_pred))
```

```python
# Decision function
decision_score=svm.decision_function(X_test)
print(decision_score)
# for each class in testing data and how many score.if core is high,it is at good
```

```python
# kernel=linear queue=c
c=np.linspace(0.001,5,20)
dff=["ovo","ovr"]
score=list()
for i in c:
    for j in dff:
        svm=SVC(kernel="linear",C=i,decision_function_shape=j)
        svm.fit(X_train,Y_train)
        y_pred=svm.predict(X_test)
        score.append([i,j,accuracy_score(Y_test,y_pred)])

score_df=pd.DataFrame(score,columns=["C","Type","Accuracy"])
score_df.sort_values(by="Accuracy",ascending=False,inplace=True)
print(score_df)
```

```python
# SC+linearsvm+pipline+c
from sklearn.pipeline import Pipeline
c=np.linspace(0.001,5,20)
dff=["ovo","ovr"]
score=list()
for i in tqdm(c):
    for j in dff:
        svm=SVC(kernel="linear",C=i,decision_function_shape=j)
        pipeline=Pipeline([("scaler",StandardScaler()),("svm",svm)])
        pipeline.fit(X_train,Y_train)
        y_pred=pipeline.predict(X_test)
        score.append([i,j,accuracy_score(Y_test,y_pred)])

score_df=pd.DataFrame(score,columns=["C","Type","Accuracy"]).sort_values("Accuracy",ascending=False)
print(score_df)
```

```python
from ucimlrepo import fetch_ucirepo
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
```

```python
car_evaluation=fetch_ucirepo(id=19)
```

```python
X=car_evaluation.data.features
Y=car_evaluation.data.targets["class"]
```

```python
encoder=LabelEncoder()
Y=encoder.fit_transform(Y)
encoder.classes_
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=25,stratify=Y)
```

```python
# linear_kernel+one hot encoder+pipeline
one_hot_encoder=OneHotEncoder(drop='first',sparse_output=False).set_output(transform="pandas")
c=np.linspace(0.001,5,20)
dff=["ovo","ovr"]
score=list()
for i in tqdm(c):
    for j in dff:
        svm=SVC(kernel="linear",C=i,decision_function_shape=j)
        pipeline=Pipeline([("one_hot_encoder",one_hot_encoder),("svm",svm)])
        pipeline.fit(X_train,Y_train)
        y_pred=pipeline.predict(X_test)
        score.append([i,j,accuracy_score(Y_test,y_pred)])

score_df=pd.DataFrame(score,columns=["C","Type","Accuracy"]).sort_values("Accuracy",ascending=False)
print(score_df)
```
