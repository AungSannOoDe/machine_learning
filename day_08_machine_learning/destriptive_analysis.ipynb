```python
# lda+logistic regression
df=pd.read_csv("../Datasets/Glass.csv")
```

```python
X=df.drop(columns=["Type"])
Y=df["Type"]
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=25,stratify=Y)
```

```python
scaler=StandardScaler()
X_train_scaler=scaler.fit_transform(X_train)
X_test_scaler=scaler.transform(X_test)
```

```python
LDA=LinearDiscriminantAnalysis().set_output(transform="pandas")
x_train_lda=LDA.fit_transform(X_train_scaler,Y_train)
x_test_lda=LDA.transform(X_test_scaler)

```

```python
logistic_regrtession=LogisticRegression(max_iter=500)
logistic_regrtession.fit(x_train_lda,Y_train)
y_pred=logistic_regrtession.predict(x_test_lda)
print(classification_report(Y_test,y_pred))
```

```python
from sklearn.pipeline import Pipeline
pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("lda", LinearDiscriminantAnalysis()),
    ("logistic", LogisticRegression(max_iter=500))
])
pipeline.fit(X_train, Y_train)
# Predict
y_pred = pipeline.predict(X_test)
# Evaluation
print(classification_report(Y_test, y_pred))
```
