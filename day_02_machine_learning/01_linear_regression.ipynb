```python
import  pandas as pd
from matplotlib import pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

##Hello

```python
insure=pd.read_csv("../Datasets/insure_auto.csv")
```

```python
insure.columns
```

```python
lr=LinearRegression()
```

```python
X=insure[["Home"]]
Y=["Operation_Cost"]
```

```python
poly=PolynomialFeatures(degree=3,include_bias=False).set_output(transform="pandas")
X_poly=poly.fit_transform(X)
X_poly
```

```python
poly=PolynomialFeatures(degree=2,include_bias=False).set_output(transform="pandas")
X_poly=poly.fit_transform(X)
print(X_poly)
```

```python
X=insure[["Home","Automobile"]]
Y=insure["Operating_Cost"]
```

```python
lr.fit(X_poly,Y)
print(lr.coef_)
print(lr.intercept_)
```

```python
poly=PolynomialFeatures(degree=2,include_bias=False).set_output(transform="pandas")
X_poly=poly.fit_transform(X)
print(X_poly)
```

```python
lr.fit(X_poly,Y)
print(lr.coef_)
print(lr.intercept_)
```
