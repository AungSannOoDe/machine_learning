```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, root_mean_squared_error
```

```python
df = pd.read_csv("../Datasets/Boston.csv")
print(df.head())

X = df.drop('medv', axis=1)
y = df['medv']
```

```python
# Ridge Regression
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25)
ridge = Ridge()
ridge.fit(X_train, y_train)
y_pred = ridge.predict(X_test)
mean_absolute_error(y_test, y_pred)
```

```python
# Hyperparameter Tuning (Alpha)
alphas = [0.001, 0.01, 0.1, 0.5, 1, 1.5, 2, 2.5, 5, 10]
for a in alphas:
    ridge = Ridge(alpha=a)
    ridge.fit(X_train, y_train)
    y_pred = ridge.predict(X_test)
    mean_absolute_error(y_test, y_pred)
    print(f"alpha={a} Error: {mean_absolute_error(y_test, y_pred)}")
```

```python
# Lasso Regression
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25)
lasso = Lasso()
lasso.fit(X_train, y_train)
y_pred = lasso.predict(X_test)
mean_absolute_error(y_test, y_pred)
# Output: 3.373066457654744
# regulazation stength
alphas = [0.001, 0.01, 0.1, 0.5, 1, 1.5, 2, 2.5, 5, 10]
score = []
for a in alphas:
    lasso = Lasso(alpha=a)
    lasso.fit(X_train, y_train)
    y_pred = lasso.predict(X_test)
    score.append([a, mean_absolute_error(y_test, y_pred)])

df_scores = pd.DataFrame(score, columns=['alpha', 'error'])
df_scores.sort_values('error')
print(df_scores.head())
```

```python
# Elastic Net Regression
alphas = [0.001, 0.01, 0.1, 0.5, 1, 1.5, 2, 2.5, 5, 10]
ratios = [0.001, 0.25, 0.5, 0.75, 1]
score = []

for a in alphas:
    for r in ratios:
        elastic = ElasticNet(alpha=a, l1_ratio=r)
        elastic.fit(X_train, y_train)
        y_pred = elastic.predict(X_test)
        error = mean_absolute_error(y_test, y_pred)
        score.append([a, r,error ])

df_scores = pd.DataFrame(score, columns=['alpha', 'l1_ratio', 'error'])
df_scores.sort_values('error')
```
