```python
import numpy as np
from sklearn.linear_model import  LogisticRegression
from sklearn.model_selection import KFold, cross_val_score,StratifiedKFold
import pandas as pd
```

```python
df=pd.read_csv("../Datasets/BreastCancer.csv",index_col=0)
```

```python
X=df.drop(columns=['Class'])
Y=df["Class"]
```

```python
lr=LogisticRegression()
k_fold=StratifiedKFold(n_splits=5,shuffle=True,random_state=25)
result=cross_val_score(lr,X,Y,cv=k_fold)
print("K fold result",result.mean())
```

```python
result=cross_val_score(lr,X,Y,cv=k_fold,scoring='roc_auc')
print("K fold result",result.mean())
```

```python
result = cross_val_score(lr, X, Y, cv=k_fold, scoring='neg_log_loss')
print("K fold result", result.mean())
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score

# Corrected typo from "set" to "sag"
solvers = ["lbfgs", "liblinear", "saga", "sag", "newton-cg", "newton-cholesky"]
cs = np.linspace(0.001, 4, 20)

scores = []
for s in solvers:
    for c in cs:

        lr = LogisticRegression(solver=s, C=c, random_state=25)

        # Evaluate using cross-validation
        cv_results = cross_val_score(lr, X, Y, cv=k_fold, scoring="roc_auc")

        # Append solver, C, and mean log loss
        scores.append([s, c, np.mean(cv_results)])

score_df = pd.DataFrame(scores, columns=["solver", "C", "mean"]).sort_values(by="mean", ascending=False)
print(score_df)
```

```python
# combination san tar
from sklearn.model_selection import  GridSearchCV
```

```python
params={"solver":["lbfgs", "liblinear", "saga", "sag", "newton-cg", "newton-cholesky"],"C":np.linspace(0.001, 4, 20)}
```

```python
gcv=GridSearchCV(lr,param_grid=params,cv=k_fold,scoring="roc_auc")
```

```python
gcv.fit(X,Y)
print(gcv.best_params_)
print(gcv.best_score_)
```

```python
df_cv=pd.DataFrame(gcv.cv_results_)
print(df_cv)
```

```python
params={"solver":["lbfgs", "liblinear", "saga", "sag", "newton-cg", "newton-cholesky"],"C":np.linspace(0.001, 4, 20)}
```

```python
rgcv=RandomizedSearchCV(lr,param_distributions=params,cv=k_fold,scoring="roc_auc",n_iter=20,random_state=25,verbose=3)
rgcv.fit(X,Y)
print(rgcv.best_params_)
print(rgcv.best_score_)
```
