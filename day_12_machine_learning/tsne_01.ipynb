```python
from sklearn.manifold import TSNE
import  pandas as pd
import numpy as np
import seaborn as sns
from  sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

```

```python
# neighbors ko yu tar
# depend on data points
# similar data point
```

```python
df=pd.read_csv("../Datasets/BreastCancer.csv",index_col=0)
X=df.drop(columns=['Class'])
y=df['Class']
```

```python
# pca
scaler=StandardScaler().set_output(transform="pandas")
x_scaled=scaler.fit_transform(X)
pca=PCA().set_output(transform="pandas")
x_pca=pca.fit_transform(x_scaled)
```

```python
data_pca=x_pca.copy()
data_pca["Class"]=y
```

```python
print(data_pca.head())
```

```python
sns.scatterplot(data=data_pca,x="pca0",y="pca1",hue="Class")
```

```python
scaler=StandardScaler().set_output(transform="pandas")
x_scaled=scaler.fit_transform(X)
tsme=TSNE(random_state=25).set_output(transform="pandas")
```

```python
x_tsme=tsme.fit_transform(x_scaled)
```

```python
tsme_data=x_tsme.copy()
tsme_data["Class"]=y
```

```python
import matplotlib.pyplot as plt
sns.scatterplot(data=tsme_data,x="tsne0",y="tsne1",hue="Class")
plt.title("t-SNE")
plt.show()
```

```python
df_vehicle=pd.read_csv("../Datasets/Vehicle.csv")
df_vehicle.columns
```

```python
X=df_vehicle.drop(columns=["Class"])
Y=df_vehicle["Class"]
```

```python
scaler=StandardScaler().set_output(transform="pandas")
x_scale=scaler.fit_transform(X)
```

```python
# n_com=2
# per=30
# rand=25
tsne=TSNE(n_components=2,perplexity=30,random_state=25).set_output(transform="pandas")
x_tsne=tsne.fit_transform(x_scale)
tsne_data=x_tsne.copy()
tsne_data["Class"]=Y
```

```python
sns.scatterplot(data=tsne_data,x="tsne0",y="tsne1",hue="Class")
plt.title("t-SNE for Vehicle ")
plt.show()
```
