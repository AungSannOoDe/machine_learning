```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import  seaborn as sns
from scipy import stats
```

```python
df=sns.load_dataset('titanic')
```

```python
print(np.max(df.isnull().sum()))
```

```python
 print(df.isnull().sum().nlargest(2))
```

```python
print(df.head(6))
```

```python
df.dropna(axis=1,how='all')
```

```python
sns.histplot(df["age"],color='blue',kde=True,bins=10)
plt.show()
```

```python
print(df["age"].skew())
```

```python
df["average_age_mean"]=df["age"].fillna(df["age"].mean())
```

```python
df["age"]=df["age"].fillna(df["age"].median())
```

```python
print(df[["average_age_mean","average_age_median","age"]])
```

```python
sns.histplot(df["average_age_median"],color='blue',kde=True,bins=10)
```

```python
stat, p_value = stats.shapiro(df['age'].dropna())
```

```python
print(f"Statistics: {stat:.4f}, P-value: {p_value:.6f}")

if p_value > 0.05:
    print("Data က Normal Distribution ဖြစ်နိုင်ခြေရှိပါတယ် (Fail to reject H0)")
else:
    print("Data က Normal Distribution မဟုတ်ပါဘူး (Reject H0)")
```
