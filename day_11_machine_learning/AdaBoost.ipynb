```python
import pandas as pd
import numpy as np
from numpy.ma.core import remainder
from sklearn.metrics._plot import confusion_matrix
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from tqdm import tqdm
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, f1_score, accuracy_score, log_loss, accuracy_score, r2_score
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier,AdaBoostClassifier,AdaBoostRegressor
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression,LinearRegression
from sklearn.linear_model import ElasticNet
```

```python
df=pd.read_csv("../Datasets/Sonar.csv")
```

```python
le=LabelEncoder()
X=df.drop(columns=["Class"])
Y=le.fit_transform(df["Class"])
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=25,stratify=Y)
```

```python
nb=GaussianNB()
ddc=DecisionTreeClassifier(max_depth=1,random_state=25)
lr=LogisticRegression()
```

```python
ada=AdaBoostClassifier(estimator=ddc,n_estimators=10,random_state=25)
ada.fit(X_train,Y_train)
y_pred=ada.predict(X_test)
y_pred_proba=ada.predict_proba(X_test)
print("log loss",log_loss(Y_test,y_pred_proba))
print("Classfification report",classification_report(Y_test,y_pred))
```

```python
estimators=[ddc,nb,lr]
no_of_estimators=[10,25,35,60]
scores=[]
for estimator in estimators:
    for n in no_of_estimators:
         ada=AdaBoostClassifier(estimator=estimator,n_estimators=n,random_state=25)
         ada.fit(X_train,Y_train)
         y_pred=ada.predict(X_test)
         y_pred_proba=ada.predict_proba(X_test)
         scores.append([estimator,n,log_loss(Y_test,y_pred_proba)])
score_df=pd.DataFrame(scores,columns=["estimator","n_estimators","log_loss"]).sort_values(by="log_loss",ascending=True)
print(score_df)
```

```python
df=pd.read_csv("../Datasets/Concrete_Data.csv")
```

```python
x=df.drop(columns=["Strength"])
y=df["Strength"]
X_train,X_test,Y_train,Y_test=train_test_split(x,y,test_size=0.3,random_state=25)
```

```python
ddr=DecisionTreeRegressor(max_depth=1,random_state=25)
lr=LinearRegression()
el=ElasticNet()
```

```python
no_of_estimators=[10,25,30,50]
estimators=[ddr,lr,el]
scores=[]
for estimator in estimators:
    for n in no_of_estimators:
        ada=AdaBoostRegressor(n_estimators=n,random_state=25,estimator=estimator)
        ada.fit(X_train,Y_train)
        y_pred=ada.predict(X_test)
        scores.append([estimator,n,r2_score(Y_test,y_pred)])
score_df=pd.DataFrame(scores,columns=["estimator","n_estimators","r2_score"]).sort_values(by="r2_score",ascending=False)
print(score_df)
```
